```python
fuels = pd.unique(df.primary_fuel)
fig, axs = plt.subplots(len(fuels), 1, sharey=True, tight_layout=True,figsize=(10, 10))

# We can set the number of bins with the *bins* keyword argument.
for i, fuel in enumerate(fuels):
    age = current_year - df[df.primary_fuel == fuel].commissioning_year
    axs[i].hist(age)
    axs[i].set_title("Hist of age for "+str(fuel))
```
